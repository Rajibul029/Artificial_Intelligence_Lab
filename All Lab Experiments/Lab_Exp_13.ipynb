```python
# ==========================================
# 1. IMPORT LIBRARIES
# ==========================================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, BatchNormalization, Activation,
    MaxPooling2D, GlobalMaxPooling2D,
    GlobalAveragePooling2D, Dense, Dropout, Concatenate
)
from tensorflow.keras.callbacks import (
    EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
)
from tensorflow.keras.utils import to_categorical

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report
)


# ==========================================
# 2. LOAD FASHION-MNIST DATASET
# ==========================================

SEED = 42
tf.keras.utils.set_random_seed(SEED)

(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

class_names = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]

num_classes = len(class_names)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
print("Number of Classes:", num_classes)


# ==========================================
# 3. DATA PREPROCESSING
# ==========================================

# Normalize pixel values
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Reshape images for 2D CNN
X_train = np.expand_dims(X_train, axis=-1)
X_test = np.expand_dims(X_test, axis=-1)

# One-hot encoding
y_train_encoded = to_categorical(y_train, num_classes)
y_test_encoded = to_categorical(y_test, num_classes)

# Training and validation split
X_train, X_val, y_train_encoded, y_val = train_test_split(
    X_train,
    y_train_encoded,
    test_size=0.15,
    random_state=SEED,
    stratify=y_train
)

print("\nTraining Shape:", X_train.shape)
print("Validation Shape:", X_val.shape)
print("Testing Shape:", X_test.shape)


# ==========================================
# 4. BUILD 2D CNN MODEL
# ==========================================

inputs = Input(shape=(28, 28, 1))

# Convolutional Block 1
x = Conv2D(32, (3, 3), padding="same", use_bias=False)(inputs)
x = BatchNormalization()(x)
x = Activation("relu")(x)
x = MaxPooling2D((2, 2))(x)

# Convolutional Block 2
x = Conv2D(64, (3, 3), padding="same", use_bias=False)(x)
x = BatchNormalization()(x)
x = Activation("relu")(x)
x = MaxPooling2D((2, 2))(x)

# Convolutional Block 3
x = Conv2D(128, (3, 3), padding="same", use_bias=False)(x)
x = BatchNormalization()(x)
x = Activation("relu", name="relu_3")(x)

# Feature Fusion
max_pool = GlobalMaxPooling2D()(x)
avg_pool = GlobalAveragePooling2D()(x)
x = Concatenate()([max_pool, avg_pool])

# Classification Layers
x = Dense(128, activation="relu")(x)
x = Dropout(0.3)(x)
x = Dense(64, activation="relu")(x)
x = Dropout(0.2)(x)

outputs = Dense(num_classes, activation="softmax")(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()


# ==========================================
# 5. TRAINING CALLBACKS
# ==========================================

callbacks = [
    EarlyStopping(
        monitor="val_loss",
        patience=12,
        restore_best_weights=True
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=5,
        min_lr=1e-6
    ),

    ModelCheckpoint(
        "best_2d_cnn.keras",
        monitor="val_loss",
        save_best_only=True
    )
]


# ==========================================
# 6. TRAIN MODEL
# ==========================================

history = model.fit(
    X_train,
    y_train_encoded,
    epochs=50,
    batch_size=64,
    validation_data=(X_val, y_val),
    callbacks=callbacks,
    verbose=1
)


# ==========================================
# 7. MODEL EVALUATION
# ==========================================

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test_encoded,
    verbose=0
)

y_prob = model.predict(X_test, verbose=0)
y_pred = np.argmax(y_prob, axis=1)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average="macro")
recall = recall_score(y_test, y_pred, average="macro")
f1 = f1_score(y_test, y_pred, average="macro")

print("\nFINAL RESULTS")
print("=" * 40)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")


# ==========================================
# 8. TRAINING AND VALIDATION CURVES
# ==========================================

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(history.history["accuracy"], label="Training")
axes[0].plot(history.history["val_accuracy"], label="Validation")
axes[0].set_title("Model Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].legend()
axes[0].grid()

axes[1].plot(history.history["loss"], label="Training")
axes[1].plot(history.history["val_loss"], label="Validation")
axes[1].set_title("Model Loss")
axes[1].set_xlabel("Epoch")
axes[1].legend()
axes[1].grid()

plt.tight_layout()
plt.show()


# ==========================================
# 9. CONFUSION MATRIX
# ==========================================

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("2D CNN Confusion Matrix")

plt.tight_layout()
plt.show()


# ==========================================
# 10. CLASSIFICATION REPORT
# ==========================================

print("\nCLASSIFICATION REPORT")

print(classification_report(
    y_test,
    y_pred,
    target_names=class_names,
    digits=4
))


# ==========================================
# 11. INTERMEDIATE FEATURE MAPS
# ==========================================

feature_model = Model(
    inputs=model.input,
    outputs=[
        model.get_layer("relu_3").output
    ]
)

sample = X_test[0:1]
feature_maps = feature_model.predict(sample, verbose=0)[0]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(sample[0, :, :, 0], cmap="gray")
plt.title(f"Original: {class_names[y_test[0]]}")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(feature_maps[:, :, 0], cmap="viridis")
plt.title("CNN Feature Map")
plt.axis("off")

plt.tight_layout()
plt.show()


# ==========================================
# 12. ACTUAL VS PREDICTED
# ==========================================

plt.figure(figsize=(12, 5))

plt.scatter(
    range(len(y_test)),
    y_test,
    label="Actual",
    marker="o",
    s=15
)

plt.scatter(
    range(len(y_pred)),
    y_pred,
    label="Predicted",
    marker="x",
    s=15
)

plt.yticks(range(num_classes), class_names)
plt.xlabel("Test Sample")
plt.ylabel("Fashion Category")
plt.title("Actual vs Predicted Classes")

plt.legend()
plt.grid()
plt.tight_layout()
plt.show()


# ==========================================
# 13. FINAL SUMMARY
# ==========================================

print("\nFINAL EXPERIMENT SUMMARY")
print("=" * 40)

print("Training Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))
print("Epochs Executed:", len(history.history["loss"]))

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")

print("\nExperiment Completed Successfully!")
```
